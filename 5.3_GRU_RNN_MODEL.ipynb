```python
from tensorflow.keras.layers import GRU,Dense,Embedding
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.preprocessing import sequence

import numpy as np
```

```python
vocab_size=5000
(x_train,y_train),(x_test,y_test)=imdb.load_data(num_words=vocab_size)
print(x_train[0])

print("Max length of a review:",len(max((x_train+x_test),key=len)))
print("Min length of a review:",len(min((x_train+x_test),key=len)))
```

```python
print("\n Frist rewiew as integers")
word_idx=imdb.get_word_index()
#Reveers the dictionary
word_idx={i:word for word,i  in word_idx.items()}

print([word_idx[i] for i in x_train[0]])

max_len=400
x_train=sequence.pad_sequences(x_train,maxlen=max_len)
x_test=sequence.pad_sequences(x_test,maxlen=max_len) #pad_sequences every review ki lenghth =400 kar dega
print("\n Shanpe of x_train:",x_train.shape)
print("Shape of x_test:",x_test.shape)
x_valid,y_valid=x_train[:64],y_train[:64]
x_train,y_train=x_train[64:],y_train[64:]

print("\n trainig data after validation split:",x_train.shape)
print("validation data:",x_valid.shape)
```

```python
#fixing every words embedding size to be 32
embd_len=32
#create a RNN model
GRU_model=Sequential(name="GRU")
GRU_model.add(Embedding(input_dim=vocab_size,output_dim=embd_len))

#in case of a stacked (more than one layer of RNN) use return_sequences=True

GRU_model.add(GRU(128,activation='tanh',return_sequences=False))
GRU_model.add(Dense(1,activation='sigmoid'))

#printing model summary
print(GRU_model.summary())

#compiling model
GRU_model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])

#training the model
history=GRU_model.fit(x_train,y_train,batch_size=64,epochs=5,verbose=1,validation_data=(x_valid,y_valid))

#printing model score on test data
print("\n Model score on test data")

test_loss,test_accuracy=GRU_model.evaluate(x_test,y_test,verbose=0) #runnning process dekhne ke liye verbose=0
print("GRU test loss:",test_loss)
print("GRU test accuracy:",test_accuracy)
```
