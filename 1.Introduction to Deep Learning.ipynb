# **Workflow of Deep Learning:-**
1. Data Ingestion
2. Data Cleaning & Preprocessing
3. Build Sequential Model Architecture(
  input_layer(neueons,Activation_function,input_dimension),
  hideen_layer(neueons,Activation_function),
  outPut_layer(neuron_based_on_target_data,Activation_function_based_on_target_data)
)
4. model.compile(loss_function,optimizer,metrics)
5. model.training(x_train,y_train)
6. model prediction and loss


#**Loss Function:-**
 - Difference b/w Actual value and Predicted value.
 - Loss is calculated for one data point.
 - Cost Function = Average loss for all data points.

 - - For Numerical Data
      - MSE (Mean Squared Error)
      - MAE (Mean Absolute Error)
 - - For Categorical Data
      - Binary Cross Entropy → For 2 classes
      - Sparse Categorical Cross Entropy → For more than 2 classes

#**Activation Functions:-**
 - Activation functions help the neural network learn non-linear patterns.
      - ReLU → Mostly used for numerical problems and hidden layers.
      - Sigmoid → Used for binary classification.
      - Softmax → Used for multi-class classification.

#**Optimizers:-**
- Optimizers reduce the loss function by updating the weights.
  - Common optimizers:-SGD,Momentum,NAG,AdaGrad,RMSProp,Adam
