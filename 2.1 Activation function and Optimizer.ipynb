# **Activation function:-**
An activation function is a mathematical function that decides whether a neuron should be activated or not.It transforms the weighted sum of a neuron input into an output that is passed to the next layer.

Without an activation function, a neural network behaves like a simple linear regression model, no matter how many hidden layers it has. Therefore, activation functions introduce non-linearity, allowing the network to learn complex patterns.

#Purpose of Activation Function
- Helps the model learn complex (non-linear) relationships.
- Decides whether a neuron should be activated.
- Enables the model to approximate complex functions.
- Makes hidden layers meaningful.
- Improves learning and helps solve training problems.


| Activation                  | Output Range     | Main Purpose                   | Commonly Used In                        | Major Limitation              |
| --------------------------- | ---------------- | ------------------------------ | --------------------------------------- | ----------------------------- |
| **Sigmoid**                 | 0 to 1           | Binary probability             | Binary classification output layer      | Vanishing gradient            |
| **Tanh**                    | -1 to 1          | Zero-centred activation        | Hidden layers (older networks)          | Vanishing gradient            |
| **ReLU**                    | 0 to ∞           | Fast hidden layer activation   | Most hidden layers (ANN, CNN)                      | Dead ReLU problem             |
| **Leaky ReLU**              | -∞ to ∞          | Solves Dead ReLU by allowing a small negative slope                  | Hidden layers of deep neural networks                          | Need to choose negative slope |
| **PReLU (Parametric ReLU)** | -∞ to ∞          | Learns the best negative slope | Large deep networks                     | More trainable parameters     |
| **Softmax**                 | 0 to 1 (sum = 1) | Multi-class probability        | Multi-class classification output layer | Used only in output layer     |


## 1. What is an activation function and why is it needed in a neural network?

**Answer:**
An activation function is a mathematical function that decides whether a neuron should be activated or not.

It is needed because it adds **non-linearity** to the neural network. This helps the model learn complex patterns and relationships in the data. Without an activation function, the network can learn only simple linear relationships.

---

## 2. Why can't we use only linear activation functions in every layer?

**Answer:**
If every layer uses a linear activation function, the entire neural network becomes a single linear model, no matter how many hidden layers it has.

As a result, it cannot learn complex or non-linear patterns.

---

## 3. Why is ReLU more popular than Sigmoid?

**Answer:**
ReLU is more popular because:

* It is simple and fast.
* It trains deep networks faster.
* It reduces the vanishing gradient problem.
* It is computationally efficient.

Sigmoid is slower and often suffers from the vanishing gradient problem.

---

## 4. What are the vanishing gradient and exploding gradient problems?

**Answer:**

**Vanishing Gradient:**
The gradients become very small during backpropagation. The model learns very slowly or stops learning.

**Exploding Gradient:**
The gradients become very large. This causes unstable training and the model may not converge.

---

## 5. What is the Dead ReLU problem?

**Answer:**
In ReLU, if a neuron always receives a negative input, its output becomes **0**.

After that, the neuron stops learning because its gradient is also **0**.

This is called the **Dead ReLU problem**.

---

## 6. Why is Sigmoid not used in hidden layers anymore?

**Answer:**
Sigmoid is rarely used in hidden layers because:

* It causes the vanishing gradient problem.
* Training becomes slow.
* It is not zero-centred.
* ReLU usually gives better performance.

---

## 7. What happens if ReLU is used in the output layer for classification?

**Answer:**
ReLU is usually **not suitable** for the output layer in classification.

* It cannot produce probabilities between **0 and 1**.
* It may produce values greater than **1**.
* For binary classification, **Sigmoid** is used.
* For multi-class classification, **Softmax** is used.

---

## 8. What happens if no activation function is used in a deep neural network?

**Answer:**
Without activation functions:

* The network behaves like a simple linear model.
* Hidden layers become useless.
* The model cannot learn complex patterns.
* Deep learning loses its main advantage.

---

## 9. Why do deep neural networks rarely use Tanh?

**Answer:**
Tanh is better than Sigmoid because it is zero-centred, but it still suffers from the **vanishing gradient problem**.

For deep networks, ReLU usually trains faster and performs better.

---

## 10. If your deep learning model is not learning because many neurons are giving zero output, what would you do?

**Answer:**
This is usually the **Dead ReLU problem**.

Possible solutions:

* Use **Leaky ReLU** instead of ReLU.
* Use **PReLU**.
* Reduce the learning rate.
* Use better weight initialization (such as He initialization).
* Check the input data and normalization.

These methods help neurons continue learning.



# Optimizer:-

An optimizer is an algorithm used in deep learning to update the weight and biases of a neural network in order to minimize the loss function and improve model accuracy.

## 1) SGD (Stochastic Gradient Descent):-

It update the model weight using the gradient computed from a small batch. It move the weight in the direction that reduces the loss.

### Advantages:-

- It is useful in small neural network and in simple ML problem where computational resource are limited.

### Disadvantages:-

- It is slow to converge.
- It zig-zag movement during optimization.
- It may get stuck in local minima and saddle point.



## 2) Momentum Optimizer:-

It improve SGD by considering the previous weight update instead of moving only based on the current gradient. It also remembers previous direction, allowing faster and smoother convergence.

### Disadvantage:-

- It can overshoot the minimum.
- It requires tuning of the momentum parameter.



## 3) NAG (Nesterov Accelerated Gradient):-

It is an improved version of Momentum. Instead of calculating the gradient at the current position, it first look ahead to the expected future position and then calculate the gradient.

### Advantage:-

- More accurate update.
- Faster convergence than Momentum.

### Disadvantage:-

- It is slightly complex.
- It has extra computation compared to Momentum.


# 4) AdaGrad:-

AdaGrad assigns a different learning rate to each parameter based on its past gradients. Frequently updated parameters receive smaller learning rates, while infrequently updated parameters receive larger learning rates.

### Advantages:-

- It automatically adjusts the learning rate.
- Excellent for sparse datasets.

### Disadvantages:-

- Learning rate continuously decreases.
- Training may stop too early.

# 5) RMSProp (Root Mean Square Propagation):-

It improves AdaGrad by using a moving average of recent squared gradients instead of accumulating all past gradients. This prevents the learning rate from becoming too small.

### Advantages:-

- Fast convergence.
- Stable learning.
- Prevents learning rate from shrinking too much.

### Disadvantages:-

- Requires hyperparameter tuning.
- Slightly more computationally expensive.

# 6) Adam:-

It combines the advantages of Momentum and RMSProp. It uses momentum to accelerate learning and adaptive learning rate to optimize each parameter efficiently.

### Advantages:-

- Fast convergence.
- Stable optimization.
- Requires less hyperparameter tuning.

### Disadvantages:-

- It may sometimes generalize slightly worse than SGD.
- It consumes more memory.

# 2) What is the difference b/w Loss Function, Cost Function and Optimizer?

## Loss Function:-

A loss function is a mathematical function that measure the error between the predicted output and the actual output for one training example.

**Ex:-**
MSE, Cross Entropy Loss.

## Cost Function:-

A cost function is the average of the loss values for all training examples. It measures the overall performance of the model.

**Ex:-**
MSE, Avg. Cross-Entropy Loss.

## Optimizer:-

An optimizer is an algorithm that updates the model's weights and biases using gradient to minimize the cost function and improve the model accuracy.

**Ex:-**
SGD, Adam, RMSProp, Adagrad.



**Loss Function:-** Measures the error between the predicted output and the actual output for a single training sample.

**Cost Function:-** Measures the average error across all training samples in the dataset.

**Optimizer:-** Updates the model's weight and biases to minimize the cost function and improve the model performance.



# 3) What role does an optimizer play during backpropagation?

**Backpropagation:-**
It is process of calculating gradients and sending the error backward through the neural network so the optimizer can update the weights and biases.

**Work:- Forward pass + backward pass**

1. **Forward propagation →** The model makes a prediction.

2. **Loss Calculation →** The loss function measures the prediction error.

3. **Backpropagation →** The error is propagated backward and gradients are calculated.

4. **Optimizer →** The optimizer updates the weights and biases using these gradients.

5. The process repeats until the loss becomes as small as possible.

**Wnew = Wold - η × error**



# 4) What information does an optimizer use to update model parameters?

**An optimizer uses the following information:**

- Current weights and biases
- Gradients calculated during backpropagation
- Learning rate
- Previous gradients or Momentum

Using this information, the optimizer updates the model parameters to reduce the loss.



# 5) What happens if we do not use an optimizer?

If we do not use an optimizer:-

- Weight will never change.
- Biases will never change.
- Loss will remain high.
- Accuracy will not improve.
- The model will not learn from the data.

The neural network will keep making the same prediction.


# 6) Why is SGD considered slow?

 SGD (Stochastic Gradient Descent) is considered slow because it updates the weight using only one training sample at a time.

This makes the update noisy and sometimes the optimizer moves in the wrong direction before correcting itself.

**As a result:**

- It needs many iterations.
- It may zig-zag toward the minimum.
- Training takes more time than adaptive optimizers like Adam.


# 7) What problem can NAG solve that Momentum cannot?

 Momentum updates the weights using the previous direction.

 NAG (Nesterov Accelerated Gradient) first looks ahead to where the weight will be, then calculates the gradient.

Because of this:

- NAG makes more accurate updates.
- It reduces overshooting.
- It converges faster.
