### 1. Difference between ANN and CNN

| ANN                                                         | CNN                                                        |
| ----------------------------------------------------------- | ---------------------------------------------------------- |
| ANN = Artificial Neural Network                             | CNN = Convolutional Neural Network                         |
| Mostly used for structured/tabular data                     | Mostly used for image and video data                       |
| Every neuron is connected to every neuron in the next layer | Uses convolution filters to extract features               |
| Has more parameters                                         | Has fewer parameters because weights are shared            |
| Does not automatically detect image features                | Automatically detects edges, textures, shapes, and objects |

**Interview Answer (Simple English):**

> ANN is mainly used for structured data, while CNN is designed for image data. CNN uses convolution and pooling layers to automatically extract image features, whereas ANN uses only fully connected layers.


## 2. Difference between Dense, Convolutional, Flatten, and Pooling Layers

### A) Dense Layer (Fully Connected Layer)

* Every neuron is connected to every neuron in the next layer.
* Mostly used at the end of the network.
* Performs final classification or prediction.

Example:

```
Dense(128)
Dense(10, activation='softmax')
```



### B) Convolutional Layer

* Uses filters (kernels) to detect features.
* Learns edges, textures, shapes, etc.
* Main building block of CNN.

Example:

```
Conv2D(32, (3,3))
```



### C) Pooling Layer

* Reduces image size.
* Keeps important information.
* Reduces computation and helps prevent overfitting.

Common types:

* Max Pooling
* Average Pooling

Example:

```
MaxPooling2D(pool_size=(2,2))
```


### D) Flatten Layer

* Converts a 2D feature map into a 1D vector.
* Connects CNN layers to Dense layers.

Example:

```
Flatten()
```


## 3. What do you mean by Kernel and Stride in CNN?

### Kernel (Filter)

A kernel is a small matrix that moves over the image to extract features like edges, corners, and textures.

Common sizes:

* 3×3
* 5×5
* 7×7

Example:

```
Kernel = 3×3
```



### Stride

Stride is the number of pixels the kernel moves after each operation.

Example:

* Stride = 1 → moves one pixel at a time.
* Stride = 2 → moves two pixels at a time.

Larger stride reduces output size faster.



### How do you select the best kernel and stride?

There is no fixed rule.

Common practice:

* **Kernel:** 3×3 (most commonly used because it captures features efficiently with fewer parameters)
* **Stride:** 1 (preserves more information)
* Use **Stride = 2** when you want to reduce the feature map size.

The best values are chosen by experimentation and validation performance.



## 4. What is Padding and why do we use it?

Padding means adding extra pixels (usually zeros) around the border of an image before applying convolution.

### Why do we use padding?

* Prevents loss of border information.
* Preserves image size.
* Allows deeper CNNs without shrinking the feature map too quickly.

### Types of Padding

1. **Valid Padding**

* No padding.
* Output becomes smaller.

2. **Same Padding**

* Adds zeros around the image.
* Output size remains the same (when stride = 1).

**Interview Answer (Simple English):**

> Padding means adding extra pixels around the image before convolution. It helps preserve edge information and keeps the output size from shrinking too quickly.
