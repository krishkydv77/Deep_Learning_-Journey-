##NER — Named Entity Recognition

- Named Entity Recognition (NER) is on NLP technique used to find important names and information in test such as people, places, organizations, dates, and amounts-and Label them with their correct type.

# In short

- NER helps a machine understand "who", "what", and "where" from text.


#NER = Entity Detection + Entity Classification

#Example (Very Simple)

Text: "Rohit Sharma Lives in Mumbai."

- NER Output

  - Ronit Sharma PERSON

   - Mumbai LOCATION (GPE)

# why Do we use NER?

- we use NER to convert unstructured text into meaningful, structured data.

#Main Reasons:-

- Automatically extract important information

- Reduce manual work in reading documents
- help machine understand context, not just keywords
- make text usable for analutics ,ml models, and datasetes

# Common Entity Types:-

| Label    | Meaning              | Example           |
| -------- | -------------------- | ----------------- |
| PERSON   | Person ka naam       | Rahul  |
| ORG      | Organization/Company | Google, Microsoft |
| GPE      | Country/City/State   | India, Delhi      |
| LOC      | Location             | Himalayas         |
| DATE     | Date                 | 20 August 2026    |
| TIME     | Time                 | 10:30 AM          |
| MONEY    | Money                | ₹50,000           |
| PERCENT  | Percentage           | 20%               |
| PRODUCT  | Product              | iPhone 17         |
| EVENT    | Event                | Olympics          |
| LAW      | Law                  | GDPR              |
| LANGUAGE | Language             | English           |
| FACILITY | Building/place       | Delhi Airport     |

```python
#ex1.
text=" Barack Obama was born in Hawaii and became President of the US"
doc=nlp(text)
for ent in doc.ents:
  print(ent.text,"-->",ent.label_)
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")

# Example 1 - Mixed entities
text = "Apple announced a partnership with Stanford University in California on 5th May 2020."
doc = nlp(text)
for ent in doc.ents:
    print(ent.text, "-->", ent.label_)
displacy.render(doc,style="ent",jupyter=True)

# Example 2 - Overlapping names
text = "Jordan visited Jordan in 2019 to watch Michael Jordan's documentary."
doc = nlp(text)
for ent in doc.ents:
    print(ent.text, "-->", ent.label_)
displacy.render(doc,style="ent",jupyter=True)

# Example 3 - Ambiguous money and date
text = "He paid 500 dollars in March, but the receipt said 03/04/2021."
doc = nlp(text)
for ent in doc.ents:
    print(ent.text, "-->", ent.label_)
displacy.render(doc,style="ent",jupyter=True)

# Example 4 - Confusing organization vs person
text = "Washington Post interviewed Washington about the Washington Monument."
doc = nlp(text)
for ent in doc.ents:
    print(ent.text, "-->", ent.label_)
displacy.render(doc,style="ent",jupyter=True)

# Example 5 - Nested entities
text = "The 2024 Olympics in Paris featured athletes sponsored by Coca-Cola."
doc = nlp(text)
for ent in doc.ents:
    print(ent.text, "-->", ent.label_)
displacy.render(doc,style="ent",jupyter=True)
```

```python
fake_review = """
Yeh restaurant Jaipur mein hai aur ambience bahut hi classy hai.
Food quality ekdum top notch, specially butter naan aur paneer tikka.
Service thodi slow thi lekin staff friendly tha.
Price thoda high laga compared to local dhabas.
Overall experience mast tha, ek baar try karna banta hai.
Last week humne birthday celebrate kiya yaha, decorations awesome the.
Music bhi soft aur relaxing tha, ekdum cafe vibes.
Parking area bada hai, so koi tension nahi.
Owner personally aake greet kiya, felt very special.
Menu mein variety hai, Indian aur Continental dono options.
Drinks section bhi acha hai, mojito aur cold coffee recommend karta hu.
Location city center mein hai, easy to reach.
Crowd decent tha, family friendly environment.
Interior design modern aur elegant hai.
Definitely worth visiting again, ek memorable outing thi.
"""

sentences = fake_review.strip().split("\n")
```

```python
!pip install googletrans==4.0.0-rc1
```

```python
from googletrans import Translator
translator = Translator()

translated_sentences = []
for s in sentences:
    result = translator.translate(s, src='hi', dest='en')
    translated_sentences.append(result.text)

print(translated_sentences)
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")

for s in translated_sentences:
    doc = nlp(s)
    for ent in doc.ents:
        print(ent.text, "-->", ent.label_)
```
