**SimpleRNN:-**
- RNN = Recurrent Neural Network.
- It reads a sentence word by word and remembers what came before.
- Useful for sequence data like text, speech, or time series.

**Dense:-**
- A fully connected layer.
- It takes the output from RNN and makes the final decision.
- For sentiment analysis, it outputs 0 or 1 (negative or positive).

**Sequential:-**
- Just a container to arrange layers in order.

                    Sequential([
                          Embedding(...),
                          SimpleRNN(...),
                          Dense(...)
                        ])

- Means: first embedding → then RNN → then dense → final result.

**Embedding:-**
- Think of it as a dictionary.
- It converts each word (which is stored as a number) into a **vector of numbers** that represent its meaning.

- **Embedding =** translate each word into a meaning.(word meanings)
- **RNN =** read the sentence step by step, remembering context.(sequence understanding)
- **Dense =** decide if the review is positive or negative.(final decision)
- **Sequential =** all steps arranged in a straight line.(pipeline to connect them)



```python
from tensorflow.keras.layers import SimpleRNN,Dense,Embedding
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
#Getting reviews with words thta come under 5000.
#most occurring words in the entire
#corpus of textual review data
vocab_size=5000
(x_train,y_train),(x_test,y_test)=imdb.load_data(num_words=vocab_size)
print(x_train[0])
```

```python
#Getting all the words from word_index dictionary
word_idx=imdb.get_word_index()

#Originally the index number of a value and not a key,
#hence converting the index as key and the words as values
word_idx={i:word for word,i  in word_idx.items()}
#agin printing the review
print([word_idx[i] for i in x_train[0]])
```

```python
#Get the minimum and the maximum length of reviews
print("Max length of a review:",len(max((x_train+x_test),key=len)))
print("Min length of a review:",len(min((x_train+x_test),key=len)))
```

```python
from tensorflow.keras.preprocessing import sequence

#keeping a fixed lenght of all reviews to max 400 words
max_len=400
x_train=sequence.pad_sequences(x_train,maxlen=max_len)
x_test=sequence.pad_sequences(x_test,maxlen=max_len)
x_valid,y_valid=x_train[:64],y_train[:64]
x_train,y_train=x_train[64:],y_train[64:]
```

```python

#fixing every words embedding size to be 32
embd_len=32
#create a RNN model
RNN_model=Sequential(name="Simple_RNN")
RNN_model.add(Embedding(vocab_size,embd_len,input_length=max_len))

#in case of a stacked (more than one layer of RNN)
#use return_sequences=True

RNN_model.add(SimpleRNN(128,activation='tanh',return_sequences=False))
RNN_model.add(Dense(1,activation='sigmoid'))

#printing model summary
print(RNN_model.summary())

#compiling model
RNN_model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])

#training the model
history=RNN_model.fit(x_train,y_train,batch_size=64,epochs=5,verbose=1,validation_data=(x_valid,y_valid))

#printing model score on test data
print()
print("Simple_RNN model score --->:",RNN_model.evaluate(x_test,y_test,verbose=0))

```
