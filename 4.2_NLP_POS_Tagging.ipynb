#**POS Tagging in NLP:-**
- POS Tagging means part of speech tagging.
- It is an NLP technique where each word in a sentence is assigned its grammatical role such as:
   - Noun
   - Verb
   - Adjective
   - Adverb
   - Pronoun
   - preposition
   - Conjunction

## why POS tagging is impoetant?
 - Pos tagging is used in:-
    - text classification
    - Chatbox
    - Machine Translation
    - Named entity recogntion(NER)
    - Sentiment Analysis
    - Speech Recognition
    - Grammar Checking

#**Common POS Tags:-**

| Tag        | Meaning        | Example         |
| ---------- | -------------- | --------------- |
| NOUN       | Noun           | book, car       |
| PROPN      | Proper Noun    | Rahul, India    |
| VERB       | Verb           | eat, play       |
| ADJ        | Adjective      | good, beautiful |
| ADV        | Adverb         | quickly, slowly |
| PRON       | Pronoun        | he, she, it     |
| DET        | Determiner     | a, an, the      |
| ADP        | Preposition    | in, on, at      |
| CONJ/CCONJ | Conjunction    | and, but        |
| NUM        | Number         | one, 10         |
| AUX        | Auxiliary Verb | is, have, can   |
| PART       | Particle       | to, not         |
| INTJ       | Interjection   | wow, oh         |

## **working Process od pos tagging:-**
 - 1. Sentence is divided into tokens.
 - 2. NLP model chack grammar/context.
 - 3. Appropriate POS tag is assigned to each word.

                          Sentence
                             ↓
                         Tokenization
                             ↓
                      Context Analysis
                             ↓
                         POS Model
                              ↓
                         POS Tags


- **POS tagging identifies the grammatical role of a word, while NER identifies the type of real-world entity.**

```python
import spacy
nlp = spacy.load('en_core_web_sm')
#imput text
doc=nlp("The smart boy quickly solved the difficult problem.")

#pos tagging
for token in doc:
  print(token.text,"----->",token.pos_)

```

```python
doc = nlp("Apple is looking at buying U.K. startup for $1 billion")
#POS tagging
for token in doc:
    print(token.text,"---->" ,token.pos_)
```

```python
# Example 1
doc = nlp("Microsoft was founded by Bill Gates in 1975.")
for token in doc:
    print(token.text,"--->", token.pos_, "--->", token.tag_)
print("===============================================")

# Example 2
doc = nlp("Sachin Tendulkar scored 100 centuries in cricket.")
for token in doc:
    print(token.text,"--->", token.pos_, "--->", token.tag_)
print("===============================================")

# Example 3
doc = nlp("The Eiffel Tower is located in Paris.")
for token in doc:
    print(token.text,"--->", token.pos_, "--->", token.tag_)
```

```python
from spacy import displacy
```

```python
# Example 4 - Ambiguous word
doc = nlp("Jordan visited Jordan in 2019.")
for token in doc:
    print(token.text,"--->", token.pos_, "--->", token.tag_)
#Display
displacy.render(doc, style="dep", jupyter=True)
print("===============================================")

# Example 5 - Money + Date confusion
doc = nlp("He paid 500 dollars in March, but the receipt said 03/04/2021.")
for token in doc:
    print(token.text,"--->", token.pos_, "--->", token.tag_)
print("===============================================")

# Example 6 - Organization vs Person
doc = nlp("Washington Post interviewed Washington about the Washington Monument.")
for token in doc:
    print(token.text,"--->", token.pos_, "--->", token.tag_)
print("===============================================")

# Example 7 - Nested entities
doc = nlp("The 2024 Olympics in Paris featured athletes sponsored by Coca-Cola.")
for token in doc:
    print(token.text,"--->", token.pos_, "--->", token.tag_)
displacy.render(doc, style="dep", jupyter=True)
```
