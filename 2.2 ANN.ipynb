```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('/content/Attrition - Attrition.csv')
df.head()
```

```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
from sklearn.metrics import accuracy_score
```

```python
df.info()
```

```python
df=df.dropna()
```

```python
lb=LabelEncoder()
for i in df.columns:
  if df[i].dtype=='object':
    df[i]=lb.fit_transform(df[i])

df.head()
```

```python
x=df.drop(columns=['Attrition'])
y=df['Attrition']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
sc=StandardScaler()
x_train_sc=sc.fit_transform(x_train)
x_test_sc=sc.transform(x_test)
```

#Now we will create architecture of our model

```python
# model=tf.keras.Sequential([
#     tf.keras.layers.Dense(128,activation='relu',input_shape=(x_train_sc.shape[1],)), #input layer
#     tf.keras.layers.Dropout(0.2),
#     tf.keras.layers.Dense(64,activation='relu'), #hidden layer
#     tf.keras.layers.Dropout(0.2),
#     tf.keras.layers.Dense(1,activation='sigmoid') #output layer
# ])


#result
# 10/10 ━━━━━━━━━━━━━━━━━━━━ 0s 5ms/step - accuracy: 0.8673 - loss: 0.3805
# Test Loss: 0.3805
# Test Accuracy: 86.73%
```

```python
model=tf.keras.Sequential([
    tf.keras.layers.Dense(64,activation='relu',input_shape=(x_train_sc.shape[1],)), #input layer
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(32,activation='relu'), #hidden layer
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(1,activation='sigmoid') #output layer
])
```

```python
#compile the model
loss_fn=tf.keras.losses.BinaryCrossentropy()
metrics=['accuracy']
```

```python
learning_rate=0.001
momentum=0.9
optimizer=tf.keras.optimizers.SGD(learning_rate=learning_rate,momentum=momentum,nesterov=True)
```

```python
model.compile(optimizer=optimizer,loss=loss_fn,metrics=metrics)
```

```python
#train the model
model.fit(x_train_sc,y_train,epochs=50,batch_size=8,validation_split=0.1)
```

```python
loss,accuracy=model.evaluate(x_test_sc,y_test)
print(f'Test Loss: {loss:.4f}')
print(f'Test Accuracy: {accuracy*100:.2f}%')
```
