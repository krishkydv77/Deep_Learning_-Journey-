#**Word2Vec:-**  dense vector representation of words trained using CBOW/Skip-gram. It captures semantic meaning and is widely used in NLP pipelines.
- A neural network–based model that converts words → vectors (embeddings).
- Words with similar meaning/context → vectors close to each other in space.

**Why Word2Vec?**
- Traditional **one-hot encoding**→ sparse, no semantic meaning.
- **TF-IDF** → captures frequency but not meaning.
- Word2Vec → captures **semantic + syntactic relationships**.

**Two Architectures:-**
- **CBOW (sg=0)**
    - Predict target word from surrounding context.
    - Faster, works well with small datasets.

- **Skip-gram (sg=1)**
    - Predict surrounding context words from target word.
    - Better for rare words, large datasets.

**Key Parameters:-**
- **vector_size** → dimension of embeddings (e.g., 50, 100, 300).

- **window** → how many words around target word are considered.

- **min_count** → ignore words that appear less than this frequency.

- **sg** → 0 = CBOW, 1 = Skip-gram.
- **workers** → number of CPU threads for training


**Using Word Vectors (wv):-**
- **Get vector**: model.wv['data']
- **Find similar words:** model.wv.most_similar('science')
- **Check similarity**: model.wv.similarity('data','science')
- **Vocabulary:** model.wv.key_to_index

**Applications:-**
- **Semantic similarity →** search engines, chatbots.
- **Text classification →** embeddings as features.
- **Clustering/Topic modeling →** group similar words.
- **Recommendation systems →** context-aware suggestions.

**Limitations:-**
- Word2Vec is **static →** one vector per word.
- Cannot handle polysemy (e.g., "bank" = river bank vs financial bank).
- Large corpus needed for good quality embeddings.

```python
from gensim.models import Word2Vec
from gensim.utils import simple_preprocess
##data
raw_corpus = [
    "Data Science uses Machine Learning",
    "Gensim library is quite useful for NLP tasks",
    "Word2Vec model gives us word embeddings",
    "Deep Learning and Neural Networks improve NLP",
    "FastText uses subword information for embeddings"
]

# Tokenization
tokenized_corpus = [simple_preprocess(doc) for doc in raw_corpus]
print(tokenized_corpus)
```

```python
#step=2 train the word2vec model
model=Word2Vec(
    sentences=tokenized_corpus, #training data (list of tokenized sentences). Model isi se word embeddings banata hai
    vector_size=50,   #dimension of embedding
    window=3, # # context window size
    min_count=1, # minimum word frequency
    sg=1, #Skip‑gram: predict context words from a given word
    workers=4 #Parallel threads for training. Matlab 4 CPU cores ek saath kaam karenge → fast training
)
```

```python
#show the vector of any word
print(model.wv["data"])
print(model.wv["data"].shape)
```

```python
#see the similar words
print(model.wv.most_similar("science",topn=3))
```

```python
#check the similarity of 2 words
print(model.wv.similarity("data","science"))
```

```python
with open("novel.txt", "r", encoding="utf-8") as f:
    raw_text = f.read()

raw_corpus = raw_text.split("\n")
tokenized_corpus = [simple_preprocess(doc) for doc in raw_corpus if doc.strip() != ""]
print(tokenized_corpus[500:700])
```

```python
model=Word2Vec(
    sentences=tokenized_corpus,
    vector_size=50,
    window=3,
    min_count=1,
    sg=1,
    workers=4
)

print(model.wv["adventures"])
print(model.wv["adventures"].shape)
```

```python
print(model.wv.most_similar("adventures",topn=10))
```

```python
print(model.wv.similarity("adventures","wonderland"))
```
