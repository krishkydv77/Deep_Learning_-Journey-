| Feature                | RNN                                  | LSTM                                        | GRU                                                   |
| ---------------------- | ------------------------------------ | ------------------------------------------- | ----------------------------------------------------- |
| Full Form              | Recurrent Neural Network             | Long Short-Term Memory                      | Gated Recurrent Unit                                  |
| Basic Idea             | Uses previous hidden state as memory | Uses gates to control long-term memory      | Uses gates to control memory with a simpler structure |
| Memory                 | Short-term memory                    | Short + long-term memory                    | Short + long-term memory                              |
| Gates                  | No gates                             | 3 gates: Forget, Input, Output              | 2 gates: Reset, Update                                |
| Cell State             |  No                                 |  Yes                                       |  No separate cell state                              |
| Hidden State           |  Yes                                |  Yes                                       |  Yes                                                 |
| Vanishing Gradient     | More affected                        | Handles it better                           | Handles it better                                     |
| Long-Term Dependencies | Poor                                 | Good                                        | Good                                                  |
| Architecture           | Simple                               | Complex                                     | Simpler than LSTM                                     |
| Parameters             | Lowest                               | Highest                                     | Lower than LSTM                                       |
| Training Speed         | Fast                                 | Slowest                                     | Faster than LSTM                                      |
| Computational Cost     | Low                                  | High                                        | Medium                                                |
| Memory Usage           | Low                                  | High                                        | Lower than LSTM                                       |
| Complexity             | Simple                               | More complex                                | Moderate                                              |
| Main Gates/Components  | Hidden state                         | Forget + Input + Output gates + Cell state  | Reset + Update gates                                  |
| Suitable For           | Simple sequential tasks              | Complex long-term dependencies              | Sequential tasks where speed and efficiency matter    |
| Example Applications   | Basic time-series, simple text tasks | Machine translation, speech, long sequences | NLP, time-series, speech, sequence modeling           |




```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, LSTM, GRU, Dense
```

```python
vocab_size=2500
(x_train,y_train),(x_test,y_test)=imdb.load_data(num_words=vocab_size)
print(x_train[0])

print("Max length of a review:",len(max((x_train+x_test),key=len)))
print("Min length of a review:",len(min((x_train+x_test),key=len)))
```

```python
print("\n Frist rewiew as integers")
word_idx=imdb.get_word_index()
#Reveers the dictionary
word_idx={i:word for word,i  in word_idx.items()}

print([word_idx[i] for i in x_train[0]])

max_len=400
x_train=sequence.pad_sequences(x_train,maxlen=max_len)
x_test=sequence.pad_sequences(x_test,maxlen=max_len)
print("\n Shanpe of x_train:",x_train.shape)
print("Shape of x_test:",x_test.shape)
x_valid,y_valid=x_train[:64],y_train[:64]
x_train,y_train=x_train[64:],y_train[64:]

print("\n trainig data after validation split:",x_train.shape)
print("validation data:",x_valid.shape)
```

```python
embd_len = 32

models_dict = {
    "SimpleRNN": Sequential([
        Embedding(vocab_size, embd_len, input_length=max_len),
        SimpleRNN(128, activation='tanh'),
        Dense(1, activation='sigmoid')
    ]),
    "LSTM": Sequential([
        Embedding(vocab_size, embd_len, input_length=max_len),
        LSTM(128, activation='tanh', return_sequences=False),
        Dense(1, activation='sigmoid')
    ]),
    "GRU": Sequential([
        Embedding(vocab_size, embd_len, input_length=max_len),
        GRU(128, activation='tanh'),
        Dense(1, activation='sigmoid')
    ])
}

results = {}
for name, model in models_dict.items():
    print(f"\n{name} model summary:")
    print(model.summary())
    model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
    history = model.fit(x_train, y_train, batch_size=64, epochs=5, verbose=1, validation_data=(x_valid, y_valid)) #Validation data ek checkpoint ha jo training ke bich bich me model evaluat karta ha ,ki kahi overfitting or overflow jayada, loss validation badne se rokta ha
    test_loss, test_acc = model.evaluate(x_test, y_test, verbose=0)
    results[name] = (test_loss, test_acc)

print("\n Final Comparison:")
for name, (loss, acc) in results.items():
    print(f"{name} → Loss: {loss:.4f}, Accuracy: {acc:.4f}")
```
